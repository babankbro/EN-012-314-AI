````python
# Knowledge Graph (ข้อมูลตัวอย่างเพื่อฝึก): เก็บความรู้เป็น triple (subject, relation, object)
triples = [
    ("มหาวิทยาลัยกาฬสินธุ์", "ตั้งอยู่ที่", "จังหวัดกาฬสินธุ์"),
    ("จังหวัดกาฬสินธุ์", "อยู่ใน", "ภาคอีสาน"),
    ("สาขาวิศวกรรมคอมพิวเตอร์", "สังกัด", "มหาวิทยาลัยกาฬสินธุ์"),
    ("EN-012-314", "เปิดสอนโดย", "สาขาวิศวกรรมคอมพิวเตอร์"),
    ("EN-012-314", "is-a", "วิชาเฉพาะด้าน"),
    ("วิชาเฉพาะด้าน", "is-a", "วิชาในหลักสูตร"),
]
# TODO: เพิ่ม triple อย่างน้อย 2 ข้อ เช่น วิชาบังคับก่อน และความสัมพันธ์ is-a ของวิชาอื่น
triples += []

def query(s=None, r=None, o=None):
    """คืน triple ที่ตรงเงื่อนไข (None = อะไรก็ได้)"""
    return [t for t in triples
            if (s is None or t[0] == s) and (r is None or t[1] == r) and (o is None or t[2] == o)]

def all_types(x):
    """อนุมานแบบสืบทอด: ถ้า x is-a A และ A is-a B แล้ว x is-a B"""
    found, frontier = [], [x]
    while frontier:
        for _, _, parent in query(s=frontier.pop(), r="is-a"):
            if parent not in found:
                found.append(parent)
                frontier.append(parent)
    return found

def explain_path(start, goal):
    """หาเส้นทางความสัมพันธ์หลายช่วง (multi-hop) ด้วย BFS"""
    frontier, seen = deque([(start, [])]), {start}
    while frontier:
        node, path = frontier.popleft()
        if node == goal:
            return path
        for s, r, o in query(s=node):
            if o not in seen:
                seen.add(o)
                frontier.append((o, path + [f"{s} --{r}--> {o}"]))

print("ใครสังกัดมหาวิทยาลัยกาฬสินธุ์:", [s for s, _, _ in query(r="สังกัด", o="มหาวิทยาลัยกาฬสินธุ์")])
print("วิชาบังคับก่อนของ EN-012-314:", [s for s, _, _ in query(r="เป็นวิชาบังคับก่อนของ", o="EN-012-314")])
print("EN-012-314 เป็นอะไรบ้าง:", all_types("EN-012-314"))
for step in explain_path("EN-012-314", "ภาคอีสาน"):
    print("  ", step)
print("จำนวน triple:", len(triples))
````

## ข้อที่ 4: การวางแผนเส้นทางด้วยการค้นหาในพื้นที่สถานะ

### โจทย์

หุ่นยนต์ส่งเอกสารต้องเดินจากจุด `S` ไปห้อง `G` บนแผนผังชั้น 3 ของอาคาร (`#` = ผนัง) เดินได้ทีละช่อง บน/ล่าง/ซ้าย/ขวา

1. ระบุองค์ประกอบของปัญหา: **initial state, actions, transition model, goal test, path cost** (เขียนในเซลล์ Markdown)
2. รันโค้ด BFS แล้วอ่านเส้นทาง (`*`)
3. **TODO:** ทางเดินช่อง (แถว 0, คอลัมน์ 5) ปิดซ่อม — ใส่พิกัดลงในลิสต์ `blocked` แล้วรันใหม่ เปรียบเทียบจำนวนก้าว

### โค้ดเริ่มต้น

````python
# แผนผังชั้น 3 อาคารคณะ (S = จุดเริ่มของหุ่นยนต์ส่งเอกสาร, G = ห้องเป้าหมาย, # = ผนัง)
floor_map = [
    "S.........",
    ".#.#.####.",
    ".#...#....",
    ".####.#.#.",
    "........#G",
]
# TODO: ทางเดินช่อง (แถว 0, คอลัมน์ 5) ปิดซ่อม — ใส่พิกัด (row, col) ที่ต้องปิดลงในลิสต์
blocked = []

grid = [list(row) for row in floor_map]
for r, c in blocked:
    grid[r][c] = "#"
start = next((r, c) for r, row in enumerate(grid) for c, ch in enumerate(row) if ch == "S")
goal = next((r, c) for r, row in enumerate(grid) for c, ch in enumerate(row) if ch == "G")

def actions(state):
    """การกระทำที่ทำได้: เดิน บน/ล่าง/ซ้าย/ขวา ไปช่องที่ไม่ใช่ผนัง"""
    r, c = state
    for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        nr, nc = r + dr, c + dc
        if 0 <= nr < len(grid) and 0 <= nc < len(grid[0]) and grid[nr][nc] != "#":
            yield (nr, nc)

def bfs(start, goal):
    frontier, parent, expanded = deque([start]), {start: None}, 0
    while frontier:
        state = frontier.popleft()
        expanded += 1
        if state == goal:                            # goal test
            path = []
            while state is not None:
                path.append(state)
                state = parent[state]
            return path[::-1], expanded
        for nxt in actions(state):
            if nxt not in parent:
                parent[nxt] = state
                frontier.append(nxt)
    return None, expanded

path, expanded = bfs(start, goal)
if path is None:
    print("ไม่มีเส้นทางไปถึงเป้าหมาย")
else:
    for r, c in path[1:-1]:
        grid[r][c] = "*"
    print("\n".join("".join(row) for row in grid))
    print(f"จำนวนก้าว = {len(path) - 1}, จำนวนสถานะที่ขยาย = {expanded}")
````

### โค้ดเริ่มต้น

````python
# สภาพแวดล้อมจำลอง: ค่าจากเซนเซอร์ในห้องเรียนทุก 15 นาที (เวลา, จำนวนคน, อุณหภูมิ °C)
percepts = [
    ("08:45", 0, 29), ("09:00", 35, 29), ("09:15", 38, 27), ("09:30", 0, 26),   # 09:30 พักเบรกสั้น
    ("09:45", 36, 27), ("10:00", 37, 26), ("10:15", 0, 26), ("10:30", 0, 27), ("10:45", 0, 28),
]

class SimpleReflexAgent:
    """ตัดสินใจจาก percept ปัจจุบันเท่านั้น"""
    def act(self, percept):
        _, people, temp = percept
        return {"ไฟ": people > 0, "แอร์": people > 0 and temp >= 26}

class ModelBasedAgent:
    """จำสถานะภายใน: ห้องว่างติดต่อกันกี่รอบแล้ว"""
    def __init__(self, patience=2):
        self.patience = patience
        self.empty_count = 0
        self.state = {"ไฟ": False, "แอร์": False}

    def act(self, percept):
        _, people, temp = percept
        # TODO: ปรับปรุงหน่วยความจำ — ถ้า people == 0 ให้เพิ่ม self.empty_count มิฉะนั้นรีเซ็ตเป็น 0
        # TODO: ถ้ามีคน → เปิดไฟ และเปิดแอร์เมื่อ temp >= 26
        #       ถ้าห้องว่าง "ติดต่อกัน" ถึง self.patience รอบ → ปิดทั้งไฟและแอร์
        #       ถ้าห้องว่างยังไม่ถึง patience → คงสถานะเดิม (self.state)
        self.state = {"ไฟ": people > 0, "แอร์": people > 0 and temp >= 26}   # ← ตอนนี้ยังทำงานเหมือน reflex
        return dict(self.state)

def simulate(agent):
    rows, prev = [], {"ไฟ": False, "แอร์": False}
    for p in percepts:
        action = agent.act(p)
        rows.append({"เวลา": p[0], "คน": p[1], "°C": p[2],
                     "ไฟ": "ON" if action["ไฟ"] else "-", "แอร์": "ON" if action["แอร์"] else "-"})
        rows[-1]["สลับแอร์"] = action["แอร์"] != prev["แอร์"]
        prev = action
    return pd.DataFrame(rows)

for agent in [SimpleReflexAgent(), ModelBasedAgent()]:
    log = simulate(agent)
    print(f"== {type(agent).__name__} ==")
    print(log.drop(columns="สลับแอร์").to_string(index=False))
    print(f"แอร์เปิด {(log['แอร์'] == 'ON').sum()} ช่วง, สั่งเปิด/ปิดแอร์ {log['สลับแอร์'].sum()} ครั้ง\n")
````

## ข้อที่ 7: Generative AI — การสร้างข้อความทีละ token และ Hallucination

### โจทย์

โค้ดนี้สร้าง **bigram language model** ขนาดจิ๋ว (ทายคำถัดไปจากคำก่อนหน้า 1 คำ) จากประโยคความรู้ 4 ประโยค แล้ว (1) คำนวณความน่าจะเป็นของประโยค (2) สุ่มสร้างประโยคใหม่ และติดป้ายว่าประโยคใดไม่มีในข้อมูลฝึก

1. รันโค้ดแล้วคำนวณด้วยมือว่าทำไม $P(\text{GPT ใช้ ส่วน Decoder})$ ได้ 0.125 (แสดงความน่าจะเป็นทีละคู่คำ)
2. **TODO:** เพิ่มประโยค `"<s> GPT ใช้ ส่วน Decoder </s>"` อีก 2 ครั้ง แล้วรันใหม่ อธิบายว่าความน่าจะเป็นของสองประโยคเปลี่ยนไปอย่างไร และเพราะอะไร
3. ชี้ประโยคที่โมเดลสร้างขึ้นซึ่ง **ผิดข้อเท็จจริง** และอธิบายว่าทำไมโมเดลจึงสร้างประโยคเหล่านั้นได้ทั้งที่ไม่เคยเห็น

### โค้ดเริ่มต้น

````python
corpus = [
    "<s> GPT ใช้ ส่วน Decoder </s>",
    "<s> BERT ใช้ ส่วน Encoder </s>",
    "<s> CNN เหมาะกับ ข้อมูล ภาพ </s>",
    "<s> LSTM เหมาะกับ ข้อมูล ลำดับ </s>",
]
# TODO: เพิ่มประโยค "<s> GPT ใช้ ส่วน Decoder </s>" อีก 2 ครั้ง แล้วสังเกตว่าความน่าจะเป็นเปลี่ยนอย่างไร
corpus += []

counts = defaultdict(Counter)
for sentence in corpus:
    words = sentence.split()
    for prev, nxt in zip(words, words[1:]):
        counts[prev][nxt] += 1

def sentence_prob(sentence):
    """P(ประโยค) = ผลคูณของ P(คำถัดไป | คำก่อนหน้า) ทุกคู่"""
    words, p = sentence.split(), 1.0
    for prev, nxt in zip(words, words[1:]):
        total = sum(counts[prev].values())
        p *= counts[prev][nxt] / total if total else 0.0
    return p

for s in ["<s> GPT ใช้ ส่วน Decoder </s>", "<s> GPT ใช้ ส่วน Encoder </s>"]:
    print(f"P({s}) = {sentence_prob(s):.3f}")

rng = random.Random(2569)
print("\nประโยคที่โมเดลสร้าง:")
for _ in range(6):
    word, out = "<s>", ["<s>"]
    while word != "</s>":
        nxt = counts[word]
        word = rng.choices(list(nxt), weights=list(nxt.values()))[0]
        out.append(word)
    sentence = " ".join(out)
    tag = "มีในข้อมูลฝึก" if sentence in corpus else "ประโยคใหม่ → ต้องตรวจข้อเท็จจริง!"
    print(f"  {' '.join(out[1:-1]):28s} ({tag})")
````
